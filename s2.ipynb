## Exercice 1 — Analyse dynamique des notes
Une université souhaite analyser les notes d’un nombre variable d’étudiants, avec des options de filtrage et
de présentation.
```
def analyser_notes(*notes, **options):
 pass

```
### Travail demandé
1. Recevoir un nombre variable de notes avec *args.
2. Utiliser lambda et filter() pour éliminer les notes en dehors de [0, 20].
3. Utiliser lambda et filter() pour identifier les notes admises.
4. Calculer la moyenne, le minimum et le maximum des notes valides.
5. Lire avec **kwargs les options seuil (10 par défaut), tri (« asc » ou « desc ») et afficher_moyenne (True par
défaut).
6. Trier les notes avec sorted() et une expression lambda.
7. Gérer le cas où aucune note valide n’est disponible.
```
analyser_notes(12, 8, 17, 21, -3, 15, 9,
 seuil=10, tri="desc", afficher_moyenne=True)
Notes valides : [17, 15, 12, 9, 8]
Moyenne : 12.20
Admis : 3 | Non admis : 2
Note maximale : 17 | Note minimale : 8
```

````python
def analyser_notes(*notes, **options):
    seuil=options.get("seuil",10)

    notefilter=list(filter(lambda x: 0<=x<=20, notes))
    admises=list(filter(lambda x: x>=10, notefilter))
    tri = sorted(notefilter,key=lambda n:n, reverse=(options.get("tri","asc")=="desc"))
    print(f"Notes Valides = {tri}")
    print(f"moyenne = {sum(notefilter)/len(notefilter)}")
    print(f"Admin = {len(admises)}",end="|")
    print(f"No admin = {len(notefilter)-len(admises)}")
    print(f"max = {max(notefilter)}",end="|")
    print(f"min = {min(notefilter)}")
analyser_notes(12, 8, 17, 21, -3, 15, 9,seuil=10, tri="desc", afficher_moyenne=True)
````

# Exercice 3 — Facturation intelligente
Une entreprise souhaite automatiser une facturation multi-produits avec validation, tri et options
commerciales.
```
def generer_facture(*produits, **options):
 pass

```
Format d’un produit : (nom, prix_unitaire, quantite)
## Travail demandé
1. Recevoir un nombre variable de produits avec *args.
2. Utiliser lambda avec filter() pour exclure les lignes dont le prix ou la quantité est non positif.
3. Utiliser lambda avec map() pour calculer les montants par produit.
4. Utiliser lambda avec sorted() pour trier les lignes par montant décroissant.
5. Gérer par **kwargs les options remise (0 %), tva (19 %), livraison (0 DT), client (« Client inconnu »).
6. Calculer le total HT, la remise, le HT après remise, la TVA et le total TTC, puis retourner un dictionnaire
récapitulatif.
```
generer_facture(
 ("Clavier", 45, 2), ("Souris", 25, 1),
 ("Ecran", 450, 1), ("Cable", -5, 3),
 remise=10, tva=19, livraison=15,
 client="ITEAM University"
)
```
Client : ITEAM University

Ecran : 450.00 DT | Clavier : 90.00 DT | Souris : 25.00 DT

Produits invalides ignores : 1

Total HT : 565.00 DT | Remise : 56.50 DT

HT apres remise : 508.50 DT | TVA : 96.62 DT

Livraison : 15.00 DT | TOTAL TTC : 620.12 DT

Bonus : rejeter explicitement les remises hors de [0, 100] et les valeurs de TVA ou de livraison négatives

````python
def generer_facture(*produits, **options):
    length = len(produits)
    valides_produits = list(filter(lambda produit : produit[1]>0 and produit[2]> 0 ,produits))
    montants = list(map(lambda produit : produit[1]*produit[2],valides_produits))
    tri= list(sorted(valides_produits,key =lambda produit : produit[1]*produit[2],reverse=True))
    print(f"clinet : {options.get('client','Client inconnu')}")
    for produit in tri:
        print(f"{produit[0]} : {produit[1]*produit[2]} DT", end =" | ")
    print()
    print(f"Produits invalides ignores : {length-len(valides_produits)}")
    print(f"Total HT : {sum(montants)} DT | Remise : {sum(montants)*(options.get('remise',0)/100)}")
    print(f"HT apres remise : {sum(montants)-(sum(montants)*(options.get('remise',0)/100))} DT | TVA : {(sum(montants)-(sum(montants)*(options.get('remise',0)/100)))*(options.get('tva',19)/100)} DT")
    print(f"livraison : {options.get('livraison',0)} DT | TOTAL TTC : {(sum(montants)-(sum(montants)*(options.get('remise',0)/100)))*(options.get('tva',19)/100)+options.get('livraison',0)+(sum(montants)-(sum(montants)*(options.get('remise',0)/100)))} DT")
    

generer_facture(
 ("Clavier", 45, 2), ("Souris", 25, 1),
 ("Ecran", 450, 1), ("Cable", -5, 3),
 remise=10, tva=19, livraison=15,
 client="ITEAM University"
)
````
