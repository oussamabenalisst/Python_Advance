```python
num = [10,20,30]
it = iter(num)
while True:
    try : 
        item = next(it)
    except StopIteration:
        break
    print(item , end = " ")
```

```python
def count_up_to(n):
    for i in range(1, n + 1):
        yield i
```

```python
gen=count_up_to(5)
```

```python
print(next(gen))
```

```python
def demo():
    print("start")
    yield "A"
    print("Resume")
    yield "B"

g= demo()
print(next(g))
print(next(g))
```
