```python
def process_sales(sales) :
    it = iter(sales)
    print("id | name | amount | status")
    for i in enumerate(sales):
        item = next(it)
        print(f"{i[0]}| {item[0]} |  {item[1]} |  {item[2]} | {item[3]}")

def high_value_sales(sales, threshold):
    for i in enumerate(sales) :
        if i[1][2] > threshold and i[1][3] == "Paid" :
            yield sales[i[0]]

def calculate_statistics(*amounts):
    if len(amounts) != 0 : 
        return (sum(amounts) , sum(amounts)/len(amounts) , max(amounts),min(amounts))
    else : 
        return("error" , "not found" , "not found" , "not found")
    
def generate_report(sales, threshold, **options) : 
    print(f"Report Title: {options.get('titre', 'Sales Report')}")
    print(f"Report Currency: {options.get('devise', 'USD')}")
    Paid = list(filter(lambda status : status[3]== "Paid",sales))
    #donne la somme de  amount 
    total_amount = sum(item[2] for item in Paid)
    print(f"Total amount of Paid sales: {total_amount}")
    print(f"Sales with status 'Paid': {Paid}")
    statistics=calculate_statistics(next(high_value_sales(sales, threshold))[2])
    uniques = list(set(map(lambda client : client[1],sales)))
    print(f"Unique clients: {uniques}")
    tristatus = sorted(sales,key=lambda x : x[2])
    print(f"Sales sorted by amount ({tristatus}):")
    print(f"le montant élevé est : {statistics[2]} et le montant le plus bas est : {statistics[3]}")
    Cancelled = (item for item in sales if item[3] == 'Cancelled')
    print(f"Sales with status 'Cancelled': {list(Cancelled)}")



sales = [
    # id  , name   , amount, status
    ("S01", "Alice", 120, "Paid"),
    ("S02", "Bob", 45, "Pending"),
    ("S03", "Alice", 230, "Paid"),
    ("S04", "Diana", 70, "Paid"),
    ("S05", "Bob", 180, "Paid"),
    ("S06", "Charlie", 95, "Cancelled"),
    ("S07", "Diana", 310, "Paid")
]
threshold = 100
# process_sales(sales)
# print(next(high_value_sales(sales, threshold)))
generate_report(sales, threshold ,devise="USD" ,titre="Sales Report")
```
