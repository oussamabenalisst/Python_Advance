### Contexte

Une entreprise de vente au détail souhaite développer une application Python pour analyser ses transactions
commerciales. Le système doit traiter les transactions, identifier les ventes de montant élevé, calculer des
statistiques et générer un rapport de synthèse.

---
Chaque transaction contient un identifiant de vente, un nom de client, un montant en TND et un statut de paiement : Paid
(payé), Pending (en attente) ou Cancelled (annulé).

---


### 1.  process_sales(sales)

Développez une fonction qui traite les ventes séquentiellement et affiche chaque transaction avec sa position. La
fonction doit également gérer correctement la fin de la séquence.

À utiliser : ```iter(), next(), enumerate(), for, StopIteration.```

````python
def process_sales(sales) :
    it = iter(sales)
    print("id | name | amount | status")
    for i in enumerate(sales):
        item = next(it)
        print(f"{i[0]}| {item[0]} |  {item[1]} |  {item[2]} | {item[3]}")
````

### 2.  high_value_sales(sales, threshold)
Développez une fonction qui produit uniquement les transactions payées dont le montant dépasse le seuil
indiqué, sans stocker toutes les transactions correspondantes dans une nouvelle liste.

À utiliser : ```yield, instructions conditionnelles, boucles, expressions génératrices```

````python
def high_value_sales(sales, threshold):
    yield from (item for item in sales if item[2] > threshold and item[3] == "Paid")
````

### 3. calculate_statistics(*amounts)
Développez une fonction qui accepte un nombre quelconque de montants et renvoie leur total, leur moyenne,
leur minimum et leur maximum. Gérez le cas où aucune valeur n’est fournie.

À utiliser : ```*args, sum(), min(), max(), instructions conditionnelles.```

````python
def calculate_statistics(*amounts):
    if len(amounts) != 0 : 
        return (sum(amounts) , sum(amounts)/len(amounts) , max(amounts),min(amounts))
    else : 
        return("error" , "not found" , "not found" , "not found")
````

### 4. generate_report(sales, threshold, **options)
Développez une fonction qui présente le nombre de transactions, le chiffre d’affaires total des ventes payées, les
clients uniques, la transaction payée de montant le plus élevé, les ventes payées triées par montant, le nombre
de ventes de montant élevé et la présence éventuelle d’une transaction annulée. Prévoyez des options, telles que
la devise et le titre du rapport.

À utiliser : ```**kwargs, filter(), map(), lambda, sorted(), any(), set() et les fonctions précédemment implémentées```

````python
def generate_report(sales, threshold, **options) : 
    print(f"Report Title: {options.get('titre', 'Sales Report')}")
    print(f"Report Currency: {options.get('devise', 'TND')}")
    print("-" * 40)
    Paid = list(filter(lambda status : status[3]== "Paid",sales))
    total_amount = sum(item[2] for item in Paid)
    print(f"Total amount of Paid sales: {total_amount}")
    print(f"Sales with status 'Paid':")
    print("-" * 40)
    process_sales(Paid)
    amounts = [i[2] for i in high_value_sales(sales, threshold)]
    statistics=calculate_statistics(*amounts)
    uniques = list(set(map(lambda client : client[1],sales)))
    print("-" * 40)
    print(f"Unique clients: {uniques}")
    print("-" * 40)
    tristatus = sorted(sales,key=lambda x : x[2])
    print(f"Sales sorted by amount :")
    process_sales(tristatus)
    print(f"le montant élevé est : {statistics[2]} et le montant le plus bas est : {statistics[3]}")
    Cancelled = (item for item in sales if item[3] == 'Cancelled')
    print("-" * 40)
    print(f"Sales with status 'Cancelled' :")
    process_sales(list(Cancelled))
    print("-" * 40)
````
