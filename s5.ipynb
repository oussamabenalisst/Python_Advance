### Exercice 11 — Listes et recherche

• Partir de la liste : [12, 5, 18, 7, 12, 3, 20, 5].

• Afficher la liste triée, le minimum, le maximum et la somme de ses éléments.

• Créer une nouvelle liste sans doublons, en conservant l’ordre de première apparition.

• Demander une valeur et afficher toutes les positions où elle se trouve dans la liste initiale.



```python
liste= [12, 5, 18, 7, 12, 3, 20, 5]
print("Liste triée : ", sorted(liste))
print("Minimum : ", min(liste))
print("Maximum : ", max(liste))
print("Somme : ", sum(liste))


liste_sans_doublons = []
for element in liste:
    if element not in liste_sans_doublons:
        liste_sans_doublons.append(element)
print("Liste sans doublons : ", liste_sans_doublons)

valeur = int(input("Donne une valeur : "))
positions = [i for i, x in enumerate(liste) if x == valeur]
if positions:
    print("La valeur se trouve aux positions : ", positions)
else:
    print("La valeur n'est pas présente dans la liste.")
```

### Exercice 13 — Fonctions réutilisables

• Définir une fonction est_premier(n) qui retourne True si n est premier et False sinon.

• Définir une fonction nombres_premiers(a, b) qui retourne la liste des nombres premiers de l’intervalle [a, b].

• Demander les deux bornes à l’utilisateur et afficher les résultats.

• Tester notamment n = 0, n = 1, n = 2 et un nombre composé.

```python
def est_premier(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def nombres_premiers(a, b):
    return [n for n in range(a, b + 1) if est_premier(n)]

a = int(input("Donne la borne inférieure : "))
b = int(input("Donne la borne supérieure : "))

print("Les nombres premiers dans l'intervalle [", a, ",", b, "] sont :")
print(nombres_premiers(a, b))

```

• Prévoir le cas d’un étudiant sans note ainsi que celui d’un identifiant inexistant

```python
dictionnaire_etudiants = {
    "name"  : str,
    "note" : list
}
def ajouter_etudiant(t):
    name = input("Entrez le nom de l'étudiant : ")
    t.append({"name": name, "note": []})

def ajouter_note(t, nom ,n):
    for etudiant in t:
        if etudiant["name"] == nom:
            for i in range(n):
                note = float(input("Entrez la note (0-20) : "))
                while note < 0 or note > 20:
                    note = float(input("Entrez une note valide (0-20) : "))
                etudiant["note"].append(note)
            return
    print("Étudiant non trouvé.")

def calculer_moyenne(t, nom):
    for etudiant in t:
        if etudiant["name"] == nom:
            if len(etudiant["note"]) == 0:
                print("Aucune note pour cet étudiant.")
                return
            moyenne = sum(etudiant["note"]) / len(etudiant["note"])
            print(f"La moyenne de {nom} est : {moyenne:.2f}")
            return
    print("Étudiant non trouvé.")

def rechercher_etudiant(t, nom):
    for etudiant in t:
        if etudiant["name"] == nom:
            print(f"Étudiant trouvé : {etudiant}")
            return
    print("Étudiant non trouvé.")

etudiants = dictionnaire_etudiants[]
n=int(input("Combien d'étudiants voulez-vous ajouter ? "))
for _ in range(n):
    ajouter_etudiant(etudiants)
name = input("Entrez le nom de l'étudiant pour ajouter des notes : ")
n_notes = int(input("Combien de notes voulez-vous ajouter ? "))
ajouter_note(etudiants, name, n_notes)
name_moyenne = input("Entrez le nom de l'étudiant pour calculer la moyenne : ")
calculer_moyenne(etudiants, name_moyenne)

print("Liste des étudiants :")
sorted_etudiants = sorted(etudiants, key=lambda moy: sum(moy["note"]) / len(moy["note"]) if len(moy["note"]) > 0 else 0, reverse=True)
print(sorted_etudiants)

```
