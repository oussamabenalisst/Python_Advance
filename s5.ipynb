### Exercice 1 — Affichage et variables

• Créer trois variables contenant un prénom, un âge et une ville.

• Afficher une phrase complète contenant les trois informations.

• Modifier la valeur de l’âge puis afficher sa nouvelle valeur.

```python
prenom="Alice"
age=20
ville="Tunis"
print("Bonjour, je m'appelle", prenom, ", j'ai", age, "ans et j'habite à", ville, ".")
age = 55
print("Bonjour, je m'appelle", prenom, ", j'ai maintenant", age, "ans et j'habite à", ville, ".")
```

### Exercice 3 — Conversions et calculs

• Demander une durée entière exprimée en secondes.

• La convertir en heures, minutes et secondes restantes.

• Exemple : 3675 secondes correspondent à 1 heure, 1 minute et 15 secondes.

```python
duree = int(input("Donne la Durée : "))
heures = duree // 3600
minutes = (duree % 3600) // 60
secondes = duree % 60
print(duree, "secondes correspondent à", heures, "heures,", minutes, "minutes et", secondes, "secondes.")
```
