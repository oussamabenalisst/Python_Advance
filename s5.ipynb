• Demander une durée entière exprimée en secondes.

• La convertir en heures, minutes et secondes restantes.

• Exemple : 3675 secondes correspondent à 1 heure, 1 minute et 15 secondes.

```python
duree = int(input("Donne la Durée : "))
heures = duree // 3600
minutes = (duree % 3600) // 60
secondes = duree % 60
print(duree, "secondes correspondent à", heures, "heures,", minutes, "minutes et", secondes, "secondes.")
```

### Exercice 4 — Décision sur une note
• Lire une note comprise entre 0 et 20 ; refuser les notes hors intervalle.

• Afficher « Ajourné » si la note est inférieure à 10, « Passable » entre 10 et moins de 12, « Assez bien »
entre 12 et moins de 14, « Bien » entre 14 et moins de 16, et « Très bien » à partir de 16.

• Tester les valeurs limites de chaque intervalle.

```python
note = float(input("Donner une note entre 0 et 20 : "))
while note < 0 or note > 20:
    note = float(input("Donner une note entre 0 et 20 : "))

if note < 10:
    print("Ajourné")
elif note < 12:
    print("Passable")
elif note < 14:
    print("Assez bien")
elif note < 16:
    print("Bien")
else:
    print("Très bien")
```
